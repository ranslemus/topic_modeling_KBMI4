```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = '../data/combined_dataset.csv'

df = pd.read_csv(DATA_PATH)

print(df.shape)

df.head()
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
print("Duplicate Rows :", df.duplicated().sum())
print("Duplicate reviewId :", df["reviewId"].duplicated().sum())
print("Duplicate Content :", df["content"].duplicated().sum())
```

```python
df.dtypes
```

dropping unnecessary columns

```python
columns_to_drop = [
  "userImage",
  "userName",
  "replyContent",
  "repliedAt",
  "appVersion",
  "reviewCreatedVersion"
]

df = df.drop(columns=columns_to_drop)

df.head()
```

checking missing values

```python
df.isnull().sum().sort_values(ascending=False)
```

duplicate check

```python
print("Duplicate Rows :", df.duplicated().sum())
print("Duplicate reviewId :", df["reviewId"].duplicated().sum())
print("Duplicate Content :", df["content"].duplicated().sum())
```

converting "at" column

```python
df["at"] = pd.to_datetime(df["at"])

df["at"].head()
```

```python
df.to_csv(
    "cleaned_dataset.csv",
    index=False
)
```
