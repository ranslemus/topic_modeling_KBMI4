```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from pathlib import Path
```

```python
csv_test = pd.read_csv('../data/raw/BRImo_reviews.csv')
csv_test.isnull().sum()
```

```python
RAW_FOLDER = Path('../data/raw')

csv_files = sorted(RAW_FOLDER.glob('*.csv'))
print(f"found {len(csv_files)} files")

for file in csv_files:
  print(file.name)
```

Load + Filter Rating

```python
datasets = {}

for file in csv_files:
  bank_name = file.stem
  df = pd.read_csv(file)

  df = df[df["score"].isin([1, 2, 3])].copy()
  datasets[bank_name] = df

  print(f"{bank_name:<15} {len(df):>8,} reviews")
```

random sampling -> biar gak bias ke periode tertentu

```python
balanced_data = []
min_reviews = min(len(df) for df in datasets.values())

for bank, df in datasets.items():
    sampled = df.sample(
        n=min_reviews,
        random_state=42
    ).copy()

    sampled["bank"] = bank.upper()

    balanced_data.append(sampled)

combined_df = pd.concat(
    balanced_data,
    ignore_index=True
)

combined_df.head()
```
