```python
import pandas as pd
```

```python
merged_df = pd.read_csv("data/merged_reviews.csv")

print(f"Total reviews before filtering: {len(merged_df):,}")
merged_df.head()
```

Rating Distribution (Before Filtering)

```python
rating_distribution = (
    merged_df["score"]
    .value_counts()
    .sort_index()
    .rename_axis("Rating")
    .reset_index(name="Count")
)

rating_distribution
```

rating percentage

```python
rating_percentage = (
    merged_df["score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .rename_axis("Rating")
    .reset_index(name="Percentage (%)")
)

rating_percentage
```

fitler rating 1 - 3

```python
filtered_df = merged_df[merged_df["score"].isin([1, 2, 3])].copy()

print(f"Total reviews after filtering: {len(filtered_df):,}")
```

```python
summary = pd.DataFrame({
    "Dataset": ["Original", "Filtered (Rating 1-3)"],
    "Total Reviews": [
        len(merged_df),
        len(filtered_df)
    ]
})

summary
```

Rating Distribution After Filtering

```python
filtered_distribution = (
    filtered_df["score"]
    .value_counts()
    .sort_index()
    .rename_axis("Rating")
    .reset_index(name="Count")
)

filtered_distribution
```

```python
filtered_df.to_csv("data/filtered_reviews.csv", index=False)
```
