```python
from google_play_scraper import reviews, Sort
from datetime import datetime

# Fetch only reviews dated between 2023-01-01 and 2025-12-31 (inclusive).
start_date = datetime(2023, 1, 1)
end_date = datetime(2025, 12, 31, 23, 59, 59)

filtered = []
continuation_token = None
batch = 200

while True:
    batch_reviews, continuation_token = reviews(
        'id.bni.wondr',
        lang='id',
        country='id',
        sort=Sort.NEWEST,
        count=batch,
        continuation_token=continuation_token
    )
    if not batch_reviews:
        break
    stop_early = False
    for r in batch_reviews:
        at = r.get('at')
        if not at:
            continue
        dt = at if isinstance(at, datetime) else None
        if dt is None:
            try:
                dt = datetime.fromisoformat(at)
            except Exception:
                continue
        # keep only reviews inside range
        if start_date <= dt <= end_date:
            filtered.append(r)
        # since we fetch newest-first, once we hit older than start, we can stop
        if dt < start_date:
            stop_early = True
    if stop_early:
        break
    if not continuation_token:
        break

# assign back to `result` so following cells work unchanged
result = filtered
```

```python
for r in result:
    print("User:", r['userName'])
    print("Score:", r['score'])
    print("Comment:", r['content'])
    print("-" * 40)
```

```python
df = pd.DataFrame(result)
df.to_csv("Wondr_BNI_reviews.csv", 
          index=False, 
          encoding="utf-8", 
          quoting=csv.QUOTE_ALL, 
          escapechar='\\')
```
