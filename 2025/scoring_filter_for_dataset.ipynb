```python
import pandas as pd
```

```python
filtered_df = pd.read_csv("merged_reviews.csv")

print(f"Total reviews before filtering: {len(filtered_df):,}")
filtered_df.head()
```

2025 only filter

```python
filtered_df = filtered_df[
    (filtered_df["at"] >= "2025-01-01") &
    (filtered_df["at"] <= "2025-12-31")
].copy()

print(f"Total reviews in 2025: {len(filtered_df):,}")
filtered_df.head()
```

Rating Distribution (Before Filtering)

```python
rating_distribution = (
    filtered_df["score"]
    .value_counts()
    .sort_index()
    .rename_axis("Rating")
    .reset_index(name="Count")
)

rating_distribution
```

rating percentage

```python
rating_percentage = (
    filtered_df["score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .rename_axis("Rating")
    .reset_index(name="Percentage (%)")
)

rating_percentage
```

fitler rating 1 - 3

```python
filtered_df = filtered_df[filtered_df["score"].isin([1, 2, 3])].copy()

print(f"Total reviews after filtering: {len(filtered_df):,}")
```

```python
summary = pd.DataFrame({
    "Dataset": ["Original", "Filtered (Rating 1-3)"],
    "Total Reviews": [
        len(filtered_df),
        len(filtered_df)
    ]
})

summary
```

Rating Distribution After Filtering

```python
filtered_distribution = (
    filtered_df["score"]
    .value_counts()
    .sort_index()
    .rename_axis("Rating")
    .reset_index(name="Count")
)

filtered_distribution
```

```python
filtered_df.to_csv("filtered_reviews.csv", index=False)
```
