```python
import pandas as pd
```

```python
filtered_df = pd.read_csv("filtered_reviews.csv")

print(f"Total Reviews: {len(filtered_df):,}")
filtered_df.head()
```

```python
filtered_df.info()
```

```python
missing_values = (
    filtered_df
    .isnull()
    .sum()
    .reset_index()
)

missing_values.columns = ["Column", "Missing Values"]

missing_values
```

```python
missing_percentage = (
    filtered_df
    .isnull()
    .mean()
    .mul(100)
    .round(2)
    .reset_index()
)

missing_percentage.columns = ["Column", "Missing Percentage (%)"]

missing_percentage
```

```python
KEEP_COLUMNS = [
    "reviewId",
    "content",
    "score",
    "at",
    "bank"
]

filtered_df = filtered_df[KEEP_COLUMNS].copy()

filtered_df.head()
```

```python
duplicate_reviewid = filtered_df["reviewId"].duplicated().sum()

print(f"Duplicate Review IDs: {duplicate_reviewid:,}")
```

```python
duplicate_content = filtered_df["content"].duplicated().sum()

print(f"Duplicate Review Contents: {duplicate_content:,}")
```

```python
empty_reviews = (
    filtered_df["content"]
    .fillna("")
    .str.strip()
    .eq("")
    .sum()
)

print(f"Empty Reviews: {empty_reviews:,}")
```

```python
review_length = filtered_df["content"].astype(str).str.len()

review_length.describe()
```

```python
short_reviews = filtered_df[
    filtered_df["content"].astype(str).str.len() < 5
]

print(f"Short Reviews (<5 chars): {len(short_reviews):,}")

short_reviews.head(20)
```

```python
filtered_df["at"] = pd.to_datetime(filtered_df["at"])

filtered_df["year"] = filtered_df["at"].dt.year

filtered_df["year"].value_counts().sort_index()
```

```python
print("Earliest Review :", filtered_df["at"].min())
print("Latest Review   :", filtered_df["at"].max())
```

```python
rating_percentage = (
    filtered_df["score"]
    .value_counts(normalize=True)
    .sort_index()
    .mul(100)
    .round(2)
    .to_frame("Percentage (%)")
)

rating_percentage
```

final quality check

```python
print("Final Dataset Shape:")
print(filtered_df.shape)

print("\nMissing Values:")
print(filtered_df.isnull().sum())

print("\nDuplicate Contents:")
print(filtered_df["content"].duplicated().sum())
```

```python
filtered_df.to_csv("EDA_dataset.csv", index=False)
```
