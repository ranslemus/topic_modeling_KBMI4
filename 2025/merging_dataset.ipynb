```python
DATA_FOLDER = Path("data")
csv_files = sorted(DATA_FOLDER.glob("*.csv"))

print(f"Found {len(csv_files)} CSV file(s):")
for file in csv_files:
  print(f"- {file.name}")
```

```python
total = 0

for file in Path("data").glob("*.csv"):
    df = pd.read_csv(file)
    print(file.name, len(df))
    total += len(df)

print("\nExpected:", total)
```

```python
dataframes = []

for file in csv_files:
    df = pd.read_csv(file)

    # Add bank name based on filename
    # Example:
    # bca.csv -> BCA
    # bri.csv -> BRI
    # mandiri.csv -> MANDIRI
    df["bank"] = file.stem.upper()

    dataframes.append(df)

merged_df = pd.concat(dataframes, ignore_index=True)

print(f"Total Reviews : {len(merged_df):,}")
print(f"Total Columns : {merged_df.shape[1]}")
```
